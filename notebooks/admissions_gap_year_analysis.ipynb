## 1. Introduction: The Shift in Medical Training

Medical school admissions have shifted from a direct pipeline to one where gap years are increasingly normalized. What was once optional is now often treated as necessary for competitiveness.

This section examines how common gap years have become among matriculating students.
### Figure 1. Gap Year Trends Among Matriculants (presentation version)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick


df = pd.read_csv("msq_years_since_college_2021-2025.csv")


year_cols = [c for c in df.columns if c != "Years Since College"]
for c in year_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

# Define gap-year groups (1+ years since college)
gap_groups = {"1-2 years", "3-4 years", "5 or more years"}
gap_df = df[df["Years Since College"].isin(gap_groups)]

# Sum percentages by year
gap_pct_by_year = (
    gap_df[year_cols]
    .sum(axis=0)
    .rename_axis("Year")
    .reset_index(name="GapYearPct")
)

gap_pct_by_year["Year"] = gap_pct_by_year["Year"].astype(int)
gap_pct_by_year = gap_pct_by_year.sort_values("Year")

# Style
sns.set_theme(style="white", context="paper")
plt.rcParams.update({
    "figure.dpi": 200,
    "savefig.dpi": 600,
    "font.size": 10,
    "axes.titlesize": 30,
    "axes.labelsize": 20,
    "xtick.labelsize": 15,
    "ytick.labelsize": 20,
})

accent = "#C00000"
edge = "#333333"

fig, ax = plt.subplots(figsize=(7.2, 4.2))

sns.barplot(
    data=gap_pct_by_year,
    x="Year",
    y="GapYearPct",
    color=accent,
    edgecolor=edge,
    linewidth=0.8,
    ax=ax,
    errorbar=None
)

# Labels and title
ax.set_title(
    "Matriculants Who've Taken ≥ 1 Gap Year, All US Medical Schools (MD)",
    pad=8,
    fontfamily="Times New Roman",
    fontsize=13
)
ax.set_xlabel("Matriculation year", fontdict={"fontfamily": "Times New Roman"})
ax.set_ylabel("Share of Matriculants", fontdict={"fontfamily": "Times New Roman"})
ax.set_ylim(bottom=68, top=75)

ax.yaxis.set_major_formatter(mtick.PercentFormatter())

plt.tight_layout()
plt.savefig("msq_gap_years_since_college_2021-2025.png", dpi=600)
plt.show()
```

## 2. Medical Education as Public Investment
Medical education in the United States is not solely a private investment made by students. It is heavily subsidized through federal spending, research funding, and graduate medical education (GME) support.

Because public dollars finance both undergraduate and residency training, the timing of when physicians enter the workforce affects the return on that investment. Delayed entry into practice therefore has system-level implications beyond individual career timelines.

### Figure 2. Federal Healthcare Spending as a Share of Total Spending


```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import textwrap

# -----------------------------
# Shares from USAspending (FY2026 obligated shares)
# -----------------------------
shares_total = {
    "Healthcare": 0.302,          # 30.2%
    "All other spending": 0.698   # 69.8%
}

# Healthcare components as shares of TOTAL
shares_health_total = {
    "Medicare": 0.169,     # 16.9%
    "Health": 0.119,       # 11.9%
    "VA medical care": 0.0139  # 1.39%
}

# Convert to shares WITHIN Healthcare for the second panel
health_total = sum(shares_health_total.values())
shares_within_health = {k: v / health_total for k, v in shares_health_total.items()}

# -----------------------------
# Helpers
# -----------------------------
def is_dark(rgba):
    r, g, b, _ = rgba
    luminance = 0.2126*r + 0.7152*g + 0.0722*b
    return luminance < 0.55

def wrap_label(s, width=18):
    return "\n".join(textwrap.wrap(s, width=width, break_long_words=False))

def stacked_barh_journal(
    ax,
    parts_pct,
    title,
    cmap_name="Blues",
    min_inside=0.09,
    min_callout=0.03,
    wrap_width=18,
    title_size=18
):
    labels = list(parts_pct.keys())
    fracs = list(parts_pct.values())

    cmap = plt.get_cmap(cmap_name)
    if len(fracs) == 1:
        colors = [cmap(0.6)]
    else:
        positions = [0.35 + 0.45*(i/(len(fracs)-1)) for i in range(len(fracs))]
        colors = [cmap(p) for p in positions]

    left = 0.0
    segments = []
    for label, frac, color in zip(labels, fracs, colors):
        ax.barh(
            [0], [frac],
            left=left,
            height=0.68,
            color=color,
            edgecolor="white",
            linewidth=1.2
        )
        segments.append((label, frac, left, left + frac, color))
        left += frac

    # Inside labels
    for label, frac, x0, x1, color in segments:
        if frac >= min_inside:
            txt_color = "white" if is_dark(color) else "black"
            ax.text(
                (x0 + x1) / 2, 0,
                f"{wrap_label(label, wrap_width)}\n{frac:.1%}",
                ha="center", va="center",
                fontsize=12,
                fontweight="bold",
                color=txt_color
            )

    # Callouts for small segments
    small = [(label, frac, x0, x1, color) for (label, frac, x0, x1, color) in segments if frac < min_inside]
    y_positions = [0.72 if i % 2 == 0 else -0.72 for i in range(len(small))]

    for (label, frac, x0, x1, color), yoff in zip(small, y_positions):
        if frac < min_callout:
            continue

        xmid = (x0 + x1) / 2
        text = f"{wrap_label(label, wrap_width)}\n{frac:.1%}"

        ax.annotate(
            text,
            xy=(xmid, 0),
            xytext=(xmid, yoff),
            ha="center", va="center",
            fontsize=11,
            fontweight="bold",
            arrowprops=dict(arrowstyle="-", lw=1.0, color="black"),
            bbox=dict(boxstyle="round,pad=0.25", fc="white", ec="black", lw=0.8)
        )

    ax.set_xlim(0, 1)
    ax.set_ylim(-1.25, 1.25)
    ax.set_yticks([])
    ax.xaxis.set_major_formatter(mtick.PercentFormatter(1.0))
    ax.set_title(title, fontsize=title_size, pad=16)

    for spine in ["top", "right", "left"]:
        ax.spines[spine].set_visible(False)

# -----------------------------
# Style
# -----------------------------
plt.rcParams.update({
    "font.family": "Times New Roman",
    "font.size": 11
})

# -----------------------------
# Figure
# -----------------------------
fig, axes = plt.subplots(
    2, 1,
    figsize=(12.5, 6.2),
    gridspec_kw={"height_ratios": [1.0, 1.0]},
    constrained_layout=True
)

# Panel A: total share
stacked_barh_journal(
    axes[0],
    shares_total,
    "Total Federal Spending",
    min_inside=0.08,
    min_callout=0.0,
    wrap_width=20,
    title_size=20
)

# Panel B: within healthcare
stacked_barh_journal(
    axes[1],
    shares_within_health,
    "Healthcare Spending ",
    min_inside=0.10,
    min_callout=0.03,
    wrap_width=20,
    title_size=20
)

fig.savefig("federal_spending_healthcare.pdf", dpi=600, bbox_inches="tight")
plt.show()
```

## 3. Measuring the Cost: Physician-Years
Medical education is often discussed in terms of cost, access, or training capacity, but less often in terms of output. If the system is publicly financed, the relevant outcome is not just how many physicians are trained, but how much clinical service they ultimately provide.

To quantify this, physician-years are introduced as the total number of years a physician spends in clinical practice after completing training.

### Defining Physician-Years

A physician-year represents one year of direct patient care delivered by a fully trained attending physician.

Total physician-years can be approximated as:

$$
Physician\text{-}Years = Retirement\ Age - Attending\ Age
$$

Attending age is defined as the age at which a physician completes training and begins independent clinical practice. For primary care physicians, this typically occurs after medical school plus a three-year residency. Retirement age is approximated using reported median retirement ages, which are centered around 65 years.

Under this framework, any delay in entering training shifts the entire career timeline forward. Even small increases in matriculation or graduation age reduce the total number of years a physician is expected to practice.

At the individual level, this reduction may appear minor. At the system level, when applied across the national medical student population, these losses accumulate into substantial reductions in total clinical service.

Using observed and projected graduation ages, physician-years can be estimated across cohorts to evaluate how delayed entry into medical training affects workforce output over time.

## 4. Results: Declining Physician-Years
Using observed and projected graduation ages, physician-years can be estimated across cohorts. This allows for a direct evaluation of how delayed entry into medical training affects total years of clinical service.

### Figure 4. Projected Physician-Years by Graduation Cohort

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

GRAD_CSV = "graduation_age_aamc_2019_2025.csv"
RETIRE_CSV = "retirement_age_petterson_2010_2013.csv"
MAT_CSV = "matriculating_ages_2021-2025.csv"


RESIDENCY_YEARS = 3
MED_SCHOOL_YEARS = 4
HIGHLIGHT = "#D66060"

GRAD_MIDPOINTS = {
    "pct_under_24": 23.0,
    "pct_24_26": 25.0,
    "pct_27_29": 28.0,
    "pct_30_32": 31.0,
    "pct_over_32": 34.0,
}

MAT_MIDPOINTS = {
    "pct_under_20": 19.0,
    "pct_20_22": 21.0,
    "pct_23_25": 24.0,
    "pct_26_28": 27.0,
    "pct_over_28": 30.0,  # open bin assumption
}


grad = pd.read_csv(GRAD_CSV).sort_values("year")
ret = pd.read_csv(RETIRE_CSV)

# Mean graduation age from bins
missing_grad_cols = [c for c in GRAD_MIDPOINTS if c not in grad.columns]


grad["mean_grad_age"] = sum(grad[c] * GRAD_MIDPOINTS[c] for c in GRAD_MIDPOINTS) / 100.0


mat_raw = pd.read_csv(MAT_CSV)


label_map = {
    "Under 20": "pct_under_20",
    "20 through 22": "pct_20_22",
    "23 through 25": "pct_23_25",
    "26 through 28": "pct_26_28",
    "Over 28": "pct_over_28",
}

mat_raw["Age Group"] = mat_raw["Age Group"].astype(str).str.strip()
mat_raw = mat_raw[mat_raw["Age Group"].isin(label_map.keys())].copy()
mat_raw["key"] = mat_raw["Age Group"].map(label_map)

# Melt to long: year, pct
year_cols = [c for c in mat_raw.columns if c.endswith("(%)")]


mat_long = mat_raw.melt(
    id_vars=["key"],
    value_vars=year_cols,
    var_name="year",
    value_name="pct"
)

# Parse year integer from "2021 (%)"
mat_long["year"] = mat_long["year"].str.extract(r"(\d{4})").astype(int)
mat_long["pct"] = pd.to_numeric(mat_long["pct"], errors="coerce")



# Pivot back to wide by year with pct_* columns
mat = mat_long.pivot_table(index="year", columns="key", values="pct", aggfunc="first").reset_index()

missing_mat_cols = [c for c in MAT_MIDPOINTS if c not in mat.columns]

mat = mat.sort_values("year")
mat["mean_mat_age"] = sum(mat[c] * MAT_MIDPOINTS[c] for c in MAT_MIDPOINTS) / 100.0

# Project to graduation year (+4) and graduation age (+4)
proj = pd.DataFrame({
    "year": mat["year"] + MED_SCHOOL_YEARS,
    "mean_grad_age": mat["mean_mat_age"] + MED_SCHOOL_YEARS
})
proj = proj[proj["year"].between(2026, 2029)]

# -----------------------
# Combine observed + projected graduation ages
# -----------------------
grad_all = pd.concat([
    grad[["year", "mean_grad_age"]],
    proj[["year", "mean_grad_age"]]
]).sort_values("year").drop_duplicates("year")

# -----------------------
# Attending age + retirement anchor
# -----------------------


retire_age = float(ret["median_retire_age_direct"].mean())

grad_all["mean_attending_age"] = grad_all["mean_grad_age"] + RESIDENCY_YEARS
grad_all["clinical_years"] = retire_age - grad_all["mean_attending_age"]

years = grad_all["year"].astype(int).to_numpy()
retire_line = np.full_like(years, retire_age, dtype=float)

# -----------------------
# Plot
# -----------------------
fig, axes = plt.subplots(2, 1, figsize=(6.5, 7), sharex=True)
divider_style = (0, (6, 6))

# Panel A
ax = axes[0]

ax.plot(years, grad_all["mean_grad_age"],
        color="black", linestyle="-", marker="o",
        label="Graduation Age")

ax.plot(years, grad_all["mean_attending_age"],
        color="black", linestyle=":", marker="o",
        label="Attending Age (+3y)")

ax.plot(years, retire_line,
        color="black", linestyle="--",
        label="Retirement Age")

ax.fill_between(
    years,
    grad_all["mean_attending_age"],
    retire_line,
    color=HIGHLIGHT,
    alpha=0.25,
    label="Physician-Years"
)

ax.axvline(2025, color="black", linestyle=divider_style, linewidth=1.5)

ax.set_ylabel("Age (years)")
ax.set_title("A. Training Completion to Retirement")

ax.legend(
    loc="center left",
    bbox_to_anchor=(0.05, 0.35),
    frameon=True,
    facecolor="white",
    edgecolor="none"
)

# Panel B
ax2 = axes[1]

ax2.plot(years, grad_all["clinical_years"],
         color=HIGHLIGHT,
         marker="o",
         linewidth=2,
         label="Physician-Years")

ax2.axvline(2025, color="black", linestyle=divider_style, linewidth=1.5)

ax2.set_ylabel("Years in Primary Care")
ax2.set_xlabel("Graduation Year")
ax2.set_title("B. Estimated Physician-Years After Training")
ax2.set_xticks(years)

ax2.legend(loc="lower left", bbox_to_anchor=(0.05, 0.05), frameon=False)

# Global title
fig.suptitle(
    "Delayed Medical Training and Projected Clinical Service Years in Primary Care",
    fontsize=15,
    y=0.93
)

plt.tight_layout(rect=[0, 0, 1, 0.95])


plt.savefig("physician_years_primary_care.pdf", dpi=600, bbox_inches="tight")
plt.show()
```

## 5. System-Level Impact

While the decline in physician-years per individual appears small, its impact becomes substantial when applied across the national medical student population. Scaling these changes provides a clearer view of the total loss in clinical service associated with delayed entry into training.

### Translating Individual Decline to Workforce Loss

Across current U.S. medical school enrollment, a reduction of approximately 0.40 physician-years per student translates to a loss of roughly 40,000 physician-years of future clinical service.

Put differently, if average graduation age were to increase by just one additional year, the resulting loss would be equivalent to removing approximately 2,900 full primary care careers from the workforce.

These estimates highlight how small shifts in training timelines compound into large-scale reductions in workforce capacity. Because medical education is publicly financed and designed to produce future clinical service, delayed entry reduces the effective return on that investment.

The cost is not simply measured in time, but in forgone patient care, reduced workforce supply, and diminished capacity to meet healthcare demand.

The question is not only how many physicians are trained, but how efficiently the system converts training into years of clinical service.

## 6. The Admissions Bottleneck
Medical school admissions operate within a competitive, zero-sum system. As applicant numbers increase and acceptance rates remain limited, additional time becomes a tool for improving competitiveness.

This creates an environment where gap years function less as optional exploration and more as strategic responses to competition.

### Figure 5. Applicants, First-Time Applicants, and Reapplicants

## 7. Interpretation: Incentives Drive Delay

Across the analyses presented, a consistent pattern emerges: delayed entry into medical training is not simply the result of individual preference, but a response to the structure of the admissions system.

As competition increases within a fixed number of seats, applicants rationally invest additional time to improve their competitiveness. Gap years become a mechanism for accumulating credentials, rather than purely a period of exploration or growth.

Because medical education is publicly financed and physician-years represent the return on that investment, this shift carries system-level consequences. Delayed entry reduces total clinical service while simultaneously introducing barriers for applicants who cannot afford prolonged periods of low or unpaid work.

The result is a system in which incentives drive delay, extending training timelines without necessarily improving alignment with workforce needs.

Addressing these outcomes requires not only expanding capacity, but reconsidering how admissions incentives shape the timeline of entry into medicine.
